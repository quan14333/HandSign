```python
import numpy as np

def suggest_threshold(distances, name):
    distances = np.array(distances)
    median_val = np.median(distances)
    p75_val = np.percentile(distances, 75)
    mean_val = np.mean(distances)
    proposed = (median_val + p75_val) / 2

    print(f"--- {name} ---")
    print(f"Median: {median_val:.4f}")
    print(f"P75: {p75_val:.4f}")
    print(f"Mean: {mean_val:.4f}")
    print(f"Ngưỡng đề xuất: {proposed:.4f}\n")
    return proposed
left_hand=r"C:\HandSign\result3\left_hand.npy"
right_hand=r"C:\HandSign\result3\right_hand.npy"
face=r"C:\HandSign\result3\face.npy"

left_hand_data = np.load(left_hand)
right_hand_data = np.load(right_hand)   
face_data = np.load(face)
t_left = suggest_threshold(left_hand_data, "Left Hand")
t_right = suggest_threshold(right_hand_data, "Right Hand")
t_face = suggest_threshold(face_data, "Face")
```
